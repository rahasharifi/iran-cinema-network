# Phase 5: Collaboration Network Metrics & Fractional Weighting

<div class='alert alert-info'>
The core objective of this project is to analyze the collaboration network within Iranian cinema to identify potential "closed circles" or exclusive groups. 
To prevent large ensemble casts from heavily skewing the network (the Clique Artifact), we apply Fractional Weighting. Finally, we calculate advanced metrics like the Herfindahl-Hirschman Index (HHI) to measure collaboration concentration.
</div>

```python
from itertools import combinations
import pandas as pd
from math import comb
```

```python
df = pd.read_parquet('../data/processed/iran_cinema_clean.parquet')
df.head()
```

<div class='alert alert-warning'>
Step 1: Posthumous Filter
Before building the network, we must ensure logical consistency by removing posthumous credits (individuals credited in titles released after their death).
</div>

<div dir='rtl' class='alert alert-danger'>
مثلا ما فیلم های اقتباسی داریم که از روی نمایشنامه هایی مانند نمایشنامه های شکسپیر ساخته شده اند. طبیعتا شکسپیر چون مرده هیچ همکاری فعالی با سینمای ایران ندارد پس وجود افرادی از این دست در این تحلیل باعث میشه شاخص هامون با فرادی غیر واقعی و تاریخی الوده بشه 
</div>

```python
df_names_full = pd.read_csv('../data/raw/name.basics.tsv.gz', sep='\t', usecols=['nconst', 'birthYear', 'deathYear'])
df_check = df.merge(df_names_full, on='nconst', how='left')

df_check['deathYear'] = pd.to_numeric(df_check['deathYear'], errors='coerce')

# drop people who died before the film creation 
posthumous = df_check[df_check['deathYear'] < df_check['startYear']]
print(posthumous[['primaryName', 'primaryTitle', 'deathYear', 'startYear']].drop_duplicates().shape)
```

```python
invalid_nconsts = set(posthumous['nconst'])
df = df[~df['nconst'].isin(invalid_nconsts)]
print(df.shape)
```

<div class='alert alert-warning'>
Step 2: Generating Collaboration Pairs (Clique Artifact)
    
To construct the network, we extract every possible pair of collaborators for each title. 
If a movie has $n$ key crew members, a simple unweighted approach creates a massive fully connected clique, disproportionately inflating the connection strength of individuals in large casts. To solve this, we use Fractional Weighting: each edge is weighted as $1 / \binom{n}{2}$.

because films with large crews have an artifcial and Disproportionate effect on the network dendity and metrics. not because a deep collaboration came to happen but just because of combinatorics. 
</div>

```python
collaboration_pairs = []

for tconst, group in df.groupby('tconst'):
    people = group[['nconst', 'primaryName']].drop_duplicates().values.tolist()
    n = len(people)
    if n < 2:
        continue
    
    weight = 1 / comb(n, 2)  
    
    for (p1, n1), (p2, n2) in combinations(people, 2):
        collaboration_pairs.append({
            'tconst': tconst,
            'person1': p1, 'name1': n1,
            'person2': p2, 'name2': n2,
            'weight': weight
        })

df_pairs = pd.DataFrame(collaboration_pairs)
print(df_pairs.shape)
df_pairs.head()
```

```python
df_pairs.to_parquet('../data/processed/collaboration_pairs.parquet')
```

<div class='alert alert-warning'>
Step 3: Creating the Long Format
we convert the pairwise combinations into a "long format" where every collaboration is registered symmetrically for both 'person1' and 'person2'.
</div>

```python
pairs_long = pd.concat([
    df_pairs[['person1', 'name1', 'person2', 'weight', 'tconst']].rename(columns={'person1':'nconst','name1':'primaryName','person2':'collaborator'}),
    df_pairs[['person2', 'name2', 'person1', 'weight', 'tconst']].rename(columns={'person2':'nconst','name2':'primaryName','person1':'collaborator'})
])

print(pairs_long.shape)
pairs_long.head()
```

 افراد با بیشترین تعداد همکاری

```python
metrics = pairs_long.groupby(['nconst', 'primaryName']).agg(
    work_count=('tconst', 'nunique'),
    unique_collaborators=('collaborator', 'nunique'),
    total_collaborations=('collaborator', 'count')
).reset_index()

metrics.sort_values('total_collaborations', ascending=False).head(10)
```

<div class='alert alert-warning'>
Step4: HHI metric
</div>

```python
pair_counts = pairs_long.groupby(['nconst', 'primaryName', 'collaborator']).size().reset_index(name='pair_collab_count')
pair_counts.head()
```

```python
def compute_hhi(group):
    shares = group['pair_collab_count'] / group['pair_collab_count'].sum()
    hhi = (shares ** 2).sum()
    top_share = shares.max()
    return pd.Series({'hhi': hhi, 'top_collaborator_share': top_share})

concentration = pair_counts.groupby(['nconst', 'primaryName']).apply(compute_hhi).reset_index()
concentration.head()
```

<div class='alert alert-warning'>
Step 6: Career Span
We extract the first and last active years for each individual to calculate their overall career length.
</div>

```python
career = df.groupby('nconst').agg(
    max_year=('startYear', 'max'),   
    min_year=('startYear', 'min')
).reset_index()                     

career['career_length'] = career['max_year'] - career['min_year'] + 1
career.head()
```

<div class='alert alert-warning'>
Step 7: Final Metric (Activity Rate)
We merge to normalize productivity against career duration.
</div>

```python
metrics = metrics.merge(concentration[['nconst', 'hhi', 'top_collaborator_share']], on='nconst', how='left')
metrics = metrics.merge(career[['nconst', 'career_length']], on='nconst', how='left')
metrics['works_per_active_year'] = metrics['work_count'] / metrics['career_length']
```

```python
metrics.to_parquet('../data/processed/person_collaboration_metrics.parquet')
```
